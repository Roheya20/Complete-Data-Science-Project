## Table 2: Interns

The `interns` table contains applicants who were selected and participated in the programme.

Each intern is linked to their original applicant record through `applicant_id`.

### Table columns

| Column | Description |
|---|---|
| intern_id | Unique identifier for each intern (Primary Key) |
| applicant_id | Connects the intern to the applicants table (Foreign Key) |
| domain | Programme domain |
| start_date | Date the intern started the programme |
| completion_status | Indicates whether the intern Completed or Dropped Out |



```python
conn.execute("""
CREATE TABLE IF NOT EXISTS interns (
    intern_id INTEGER PRIMARY KEY,
    applicant_id INTEGER,
    domain TEXT,
    start_date TEXT,
    completion_status TEXT,
    FOREIGN KEY (applicant_id) REFERENCES applicants(applicant_id)
);
""")

print("Interns table created successfully!")
```

## Table 3: Hackathon Scores

The `hackathon_scores` table contains scores from the end-of-program hackathon for interns who reached that stage.

Each record is linked to an intern through the `intern_id` column.

### Table columns

| Column | Description |
|---|---|
| intern_id | Identifies the intern (Foreign Key) |
| score | Score achieved in the hackathon |

```python
conn.execute("""
CREATE TABLE IF NOT EXISTS hackathon_scores (
    intern_id INTEGER,
    score REAL,
    domain TEXT,
    FOREIGN KEY (intern_id) REFERENCES interns(intern_id)
);
""")

conn.commit()

print("Hackathon scores table created successfully!")
```

## Part A: SQL Queries and Analysis

### Query 1: How many interns completed each domain's programme?

This query counts the number of interns who completed the programme in each domain and displays the domains from the highest completed count to the lowest.

```python
# Counts completed interns grouped by domain
query1 = """
SELECT domain, COUNT(*) AS completed_count
FROM interns
WHERE completion_status = 'Completed'
GROUP BY domain
ORDER BY completed_count DESC;
"""

result1 = pd.read_sql_query(query1, conn)
display(result1)
```

**Interpretation:** AI had the highest number of completed interns, with 4, followed by Cybersecurity with 2. Web Development and Data Science each had 1 completed intern. Overall, 8 interns completed the programme across the four domains.

### Query 2: Average Hackathon Score per Domain

This query calculates the average hackathon score for each programme domain and ranks the domains from the highest average score to the lowest.

```python
# Calculates the average hackathon score for each domain
query2 = """
SELECT domain, ROUND(AVG(score), 2) AS avg_score
FROM hackathon_scores
GROUP BY domain
ORDER BY avg_score DESC;
"""

result2 = pd.read_sql_query(query2, conn)
display(result2)
```

**Interpretation:** Data Science achieved the highest average hackathon score of 95.00, followed by Web Development with 85.00. AI recorded an average of 81.50, while Cybersecurity had the lowest average of 77.50. These results suggest that Data Science interns performed best in the simulated hackathon dataset.

### Query 3: Identifying Top Performers

This query identifies interns who achieved a hackathon score of 85 or above. These interns can be considered top performers and potential candidates for recognition, certificates, or programme showcases.

The query combines the `interns` and `hackathon_scores` tables using their shared `intern_id` column.

```python
# Lists interns who scored 85 or above
query3 = """
SELECT i.intern_id, i.domain, h.score
FROM interns i
JOIN hackathon_scores h
    ON i.intern_id = h.intern_id
WHERE h.score >= 85
ORDER BY h.score DESC;
"""

result3 = pd.read_sql_query(query3, conn)
display(result3)
```

**Interpretation:** Four interns achieved a hackathon score of 85 or above. The highest score was 95, achieved by an intern in Data Science, followed by AI with 92, Cybersecurity with 88, and Web Development with 85. These interns are potential candidates for certificates or programme showcases based on the simulated results.

### Query 4: Conversion Rate from Applicants to Completed Interns per Domain

This query compares the total number of applicants with the number of interns who completed the programme in each domain. It calculates the conversion rate as a percentage to show how many applicants progressed to programme completion.

```python
# Calculates the percentage of applicants who completed the programme in each domain
query4 = """
SELECT
    a.domain_applied AS domain,
    COUNT(DISTINCT a.applicant_id) AS total_applicants,
    COUNT(DISTINCT i.intern_id) AS total_completed,
    ROUND(
        100.0 * COUNT(DISTINCT i.intern_id)
        / COUNT(DISTINCT a.applicant_id),
        2
    ) AS conversion_rate_pct
FROM applicants a
LEFT JOIN interns i
    ON a.applicant_id = i.applicant_id
    AND i.completion_status = 'Completed'
GROUP BY a.domain_applied
ORDER BY conversion_rate_pct DESC;
"""

result4 = pd.read_sql_query(query4, conn)
display(result4)
```

Interpretation: Cybersecurity recorded the highest completion rate at 16.67%, followed by AI at 13.79%. Web Development and Data Science each recorded 9.09%.

```python
# Counts completed and dropped-out interns in each domain
query5 = """
SELECT
    domain,
    completion_status,
    COUNT(intern_id) AS total_interns
FROM interns
GROUP BY domain, completion_status
ORDER BY domain, completion_status;
"""

result5 = pd.read_sql_query(query5, conn)
display(result5)
```

Interpretation:

The results show that AI had the highest number of interns, with 4 completing the programme and 1 dropping out. Cybersecurity recorded 2 completed interns and 1 dropout. Data Science and Web Development each had 1 intern who completed the programme, with no recorded dropouts.

Overall, AI had the highest number of completed interns, while both AI and Cybersecurity recorded dropouts. The results suggest that AI had the largest internship participation among the listed domains.
